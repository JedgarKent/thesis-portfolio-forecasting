```python
closed_full_stock_prices = full_stock_prices[["Close", "Volume"]]
```

```python
closed_full_stock_prices.head()
```

```python
csv_tickers = set(stock_table["ticker"].tolist())
```

```python
closed_full_stock_prices.columns
```

```python
notna_full_stock_prices = closed_full_stock_prices.dropna(axis=1, how="all")
```

```python
chosen_tickers = set(notna_full_stock_prices.columns.get_level_values(1).tolist())
```

```python
difference = csv_tickers - chosen_tickers
len(difference)
```

```python
sorted(difference)
```

```python
close = notna_full_stock_prices["Close"]
na_close = close.isna().any()
```

```python
partial = close.loc[:, na_close] 
partial.shape
```

```python
first_dates = partial.apply(lambda x: x.first_valid_index())
first_dates = pd.to_datetime(first_dates).sort_values()

first_dates
```

```python
first_dates.dt.year.value_counts().sort_index()
```

```python
gaps = close.apply(lambda x: x.loc[x.first_valid_index():].isna().sum())
gaps = gaps[gaps > 0].sort_values(ascending=False)

print(len(gaps))
gaps
```
