```python
import pandas as pd
import yfinance as yf
import requests
import io
print(pd.__version__)
import re
```

```python
url = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"
headers = {"User-Agent": "Mozilla/5.0 (thesis-portfolio-forecasting)"}
response = requests.get(url, headers=headers, timeout=10)

print(response.status_code)
```

```python
stocks = pd.read_html(io.StringIO(response.text))
final_table = stocks[0]
print(len(final_table))
```

```python
final_table.head()
```

```python
stock_table = final_table[["Symbol", "Security", "GICS Sector"]]
stock_table.head(100)
```

```python
stock_table = stock_table.rename(columns={"Symbol": "ticker", "Security": "name", "GICS Sector": "sector"})
stock_table.head(100)
```

```python
bad = stock_table["ticker"].str.contains(r"[@_!#$%^&*()<>?/\|}{~:.,;]")
print(bad.sum())
stock_table[bad]
```

```python
stock_table["ticker"] = stock_table["ticker"].str.replace(".", "-", regex=False)
```

```python
bad_after = stock_table["ticker"].str.contains(r"[@_!#$%^&*()<>?/\|}{~:.,;]")
print(bad_after.sum())
```
